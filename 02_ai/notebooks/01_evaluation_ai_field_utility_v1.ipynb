# CHAPTER 1. 상담 업무별 데이터 필요성 분석

## 각 상담에서 다음 두 가지를 동시에 분석한다.

1. 어떤 데이터 Field가 상담 수행에 필요한가
2. 그 데이터가 어느 수준까지 구체적으로 요구되는가

구체화 수준:

- L0 CONTEXT_ONLY
  - 계좌, 카드, 대출 등 업무 대상만 언급

- L1 FIELD_REQUIRED
  - 계좌번호, 거래내역, 생년월일 등 필요한 Field가 명시됨

- L2 PARTIAL_VALUE
  - 일부 자리, 마스킹 값 등 부분 정보 수준

- L3 EXACT_VALUE
  - 실제 금액, 금리, 잔액, 날짜, 식별값 등 정확한 값이 요구됨

이를 통해 Field별로 KEEP / MASK / HMAC / TOKEN / DROP 설계에 필요한
Utility Requirement를 도출한다.

```python
# 07. 데이터 Field 및 구체화 수준 규칙

FIELD_RULES = {

    "CUSTOMER_NAME": {
        "field": [r"고객명", r"성명", r"이름"],
        "partial": [r"[■●○*]{2,}", r"김\*", r"이\*", r"박\*"],
        "exact": []
    },

    "RESIDENT_ID": {
        "field": [r"주민등록번호", r"주민번호"],
        "partial": [
            r"\d{6}-?[1-4][*■●]{6}",
            r"\d{6}-?[*■●]{7}"
        ],
        "exact": [
            r"\d{6}-?[1-4]\d{6}"
        ]
    },

    "DATE_OF_BIRTH": {
        "field": [r"생년월일", r"출생일"],
        "partial": [],
        "exact": [
            r"\d{4}[.\-/년]\s*\d{1,2}[.\-/월]\s*\d{1,2}",
            r"\d{6}"
        ]
    },

    "PHONE_NUMBER": {
        "field": [r"전화번호", r"휴대폰\s*번호", r"연락처"],
        "partial": [
            r"01[016789]-?[*■●]{3,4}-?[*■●]{4}"
        ],
        "exact": [
            r"01[016789]-?\d{3,4}-?\d{4}"
        ]
    },

    "ACCOUNT_NUMBER": {
        "field": [r"계좌번호", r"계좌\s*번호"],
        "partial": [
            r"\d{2,6}[- ]?[*■●]{2,}[- ]?\d{2,6}",
            r"끝\s*\d+\s*자리"
        ],
        "exact": [
            r"\b\d{2,6}-\d{2,6}-\d{2,8}\b"
        ]
    },

    "CARD_NUMBER": {
        "field": [r"카드번호", r"카드\s*번호"],
        "partial": [
            r"\d{4}-?[*■●]{4}-?[*■●]{4}-?\d{4}",
            r"카드.*끝\s*4자리"
        ],
        "exact": [
            r"\b\d{4}-?\d{4}-?\d{4}-?\d{4}\b"
        ]
    },

    "TRANSACTION_HISTORY": {
        "field": [
            r"거래\s*내역", r"이체\s*내역",
            r"입출금\s*내역", r"결제\s*내역"
        ],
        "partial": [],
        "exact": []
    },

    "TRANSACTION_AMOUNT": {
        "field": [
            r"거래\s*금액", r"이체\s*금액",
            r"입금\s*금액", r"출금\s*금액",
            r"결제\s*금액"
        ],
        "partial": [],
        "exact": [
            r"\d[\d,]*\s*(?:원|만원|억원)"
        ]
    },

    "ACCOUNT_BALANCE": {
        "field": [r"잔액", r"계좌\s*잔액", r"예금\s*잔액"],
        "partial": [],
        "exact": [
            r"\d[\d,]*\s*(?:원|만원|억원)"
        ]
    },

    "INTEREST_RATE": {
        "field": [r"금리", r"이자율", r"연체\s*이자"],
        "partial": [],
        "exact": [
            r"\d+(?:\.\d+)?\s*%"
        ]
    },

    "LOAN_AMOUNT": {
        "field": [
            r"대출\s*금액", r"대출\s*잔액",
            r"대출\s*원금", r"상환\s*금액"
        ],
        "partial": [],
        "exact": [
            r"\d[\d,]*\s*(?:원|만원|억원)"
        ]
    },

    "DELINQUENCY_AMOUNT": {
        "field": [r"연체\s*금액", r"연체액"],
        "partial": [],
        "exact": [
            r"\d[\d,]*\s*(?:원|만원|억원)"
        ]
    },

    "DELINQUENCY_PERIOD": {
        "field": [r"연체\s*기간", r"연체\s*일수"],
        "partial": [],
        "exact": [
            r"\d+\s*(?:일|개월|년)"
        ]
    },

    "CREDIT_SCORE": {
        "field": [r"신용\s*점수", r"신용\s*등급"],
        "partial": [],
        "exact": [
            r"\d{3,4}\s*점"
        ]
    },

    "INCOME": {
        "field": [r"연\s*소득", r"월\s*소득", r"급여"],
        "partial": [],
        "exact": [
            r"\d[\d,]*\s*(?:원|만원|억원)"
        ]
    }
}
```

```python
# 08. 상담별 Field 구체화 수준 태깅

import re

LEVEL_SCORE = {
    "CONTEXT_ONLY": 0,
    "FIELD_REQUIRED": 1,
    "PARTIAL_VALUE": 2,
    "EXACT_VALUE": 3
}


def analyze_field_requirement(text):

    text = str(text)
    result = []

    for field, rules in FIELD_RULES.items():

        field_hit = any(
            re.search(p, text, re.I)
            for p in rules["field"]
        )

        partial_hit = any(
            re.search(p, text, re.I)
            for p in rules["partial"]
        )

        exact_hit = any(
            re.search(p, text, re.I)
            for p in rules["exact"]
        )

        if not (field_hit or partial_hit or exact_hit):
            continue

        if exact_hit:
            level = "EXACT_VALUE"

        elif partial_hit:
            level = "PARTIAL_VALUE"

        else:
            level = "FIELD_REQUIRED"

        result.append({
            "field": field,
            "level": level,
            "score": LEVEL_SCORE[level]
        })

    return result


df_clean["data_requirements"] = (
    df_clean["consulting_content"]
    .apply(analyze_field_requirement)
)
```

```python
# 09. 상담별 필요 Field와 구체화 수준 확인

df_clean["required_field_count"] = (
    df_clean["data_requirements"]
    .apply(len)
)

df_clean["max_specificity"] = (
    df_clean["data_requirements"]
    .apply(
        lambda x: max(
            [i["score"] for i in x],
            default=0
        )
    )
)

display(
    df_clean[
        [
            "consulting_content",
            "data_requirements",
            "required_field_count",
            "max_specificity"
        ]
    ].head(20)
)
```

```python
# 10. Field 주변 문맥 기반 구체화 수준 판정

def get_context(text, match, window=50):
    start = max(0, match.start() - window)
    end = min(len(text), match.end() + window)
    return text[start:end]


def analyze_field_requirement_v2(text):

    text = str(text)
    results = []

    for field, rules in FIELD_RULES.items():

        best_level = None
        best_score = -1

        for field_pattern in rules["field"]:

            for match in re.finditer(
                field_pattern,
                text,
                flags=re.IGNORECASE
            ):

                context = get_context(
                    text,
                    match,
                    window=50
                )

                level = "FIELD_REQUIRED"

                # 정확값 확인
                if any(
                    re.search(p, context, re.I)
                    for p in rules["exact"]
                ):
                    level = "EXACT_VALUE"

                # 부분값 확인
                elif any(
                    re.search(p, context, re.I)
                    for p in rules["partial"]
                ):
                    level = "PARTIAL_VALUE"

                score = LEVEL_SCORE[level]

                if score > best_score:
                    best_level = level
                    best_score = score

        if best_level is not None:
            results.append({
                "field": field,
                "level": best_level,
                "score": best_score
            })

    return results


df_clean["data_requirements_v2"] = (
    df_clean["consulting_content"]
    .apply(analyze_field_requirement_v2)
)
```

```python
# 11. Field × 구체화 수준 분석

requirements = (
    df_clean[["data_requirements_v2"]]
    .explode("data_requirements_v2")
    .dropna()
)

requirements["field"] = (
    requirements["data_requirements_v2"]
    .apply(lambda x: x["field"])
)

requirements["level"] = (
    requirements["data_requirements_v2"]
    .apply(lambda x: x["level"])
)

specificity_table = pd.crosstab(
    requirements["field"],
    requirements["level"]
)

specificity_ratio = (
    pd.crosstab(
        requirements["field"],
        requirements["level"],
        normalize="index"
    )
    * 100
).round(1)

display(specificity_table)
display(specificity_ratio)
```

## 12. Field별 Value Requirement 분석

상담에서 특정 Field가 언급되는 것과
해당 Field의 실제 원문값이 필요한 것은 구분한다.

각 Field에 대해 전체 탐지 건수 중
부분값 또는 정확값까지 관측된 비율을 Value Requirement Ratio로 계산한다.

이를 통해 식별정보와 금융업무 수치정보 간
Utility Requirement의 차이를 비교하고,
Transform Treatment 설계 근거로 활용한다.

```python
# 12. Field별 Value Requirement Ratio 계산

summary = specificity_table.copy()

for col in [
    "FIELD_REQUIRED",
    "PARTIAL_VALUE",
    "EXACT_VALUE"
]:
    if col not in summary.columns:
        summary[col] = 0

summary["TOTAL"] = (
    summary["FIELD_REQUIRED"]
    + summary["PARTIAL_VALUE"]
    + summary["EXACT_VALUE"]
)

summary["VALUE_REQUIRED"] = (
    summary["PARTIAL_VALUE"]
    + summary["EXACT_VALUE"]
)

summary["VALUE_REQUIREMENT_RATIO"] = (
    summary["VALUE_REQUIRED"]
    / summary["TOTAL"]
    * 100
).round(2)

summary["EXACT_REQUIREMENT_RATIO"] = (
    summary["EXACT_VALUE"]
    / summary["TOTAL"]
    * 100
).round(2)

display(
    summary[
        [
            "TOTAL",
            "FIELD_REQUIRED",
            "PARTIAL_VALUE",
            "EXACT_VALUE",
            "VALUE_REQUIREMENT_RATIO",
            "EXACT_REQUIREMENT_RATIO"
        ]
    ].sort_values(
        "VALUE_REQUIREMENT_RATIO",
        ascending=False
    )
)
```

## 13. 데이터 유형별 Utility Requirement 재구성

Field마다 업무상 필요한 정보 수준의 의미가 다르므로
단일 EXACT_VALUE 기준을 모든 데이터에 동일하게 적용하지 않는다.

### Identifier
이름, 주민번호, 전화번호, 계좌번호, 카드번호
→ 전체 식별값 / 부분 식별값 / 식별값 불필요

### Numeric Value
금액, 잔액, 금리, 소득, 신용점수, 연체기간
→ 정확한 수치값 필요 여부

### Record / History
거래내역
→ 단순 내역 존재 여부가 아니라 기간·거래목록·세부내역 수준의 필요성

이를 기반으로 데이터 유형별 Transform 설계를 분리한다.

```python
# 13. Field 유형 정의

field_type_map = {

    "CUSTOMER_NAME": "IDENTIFIER",
    "RESIDENT_ID": "IDENTIFIER",
    "DATE_OF_BIRTH": "IDENTIFIER",
    "PHONE_NUMBER": "IDENTIFIER",
    "ACCOUNT_NUMBER": "IDENTIFIER",
    "CARD_NUMBER": "IDENTIFIER",

    "INTEREST_RATE": "NUMERIC",
    "LOAN_AMOUNT": "NUMERIC",
    "DELINQUENCY_AMOUNT": "NUMERIC",
    "DELINQUENCY_PERIOD": "NUMERIC",
    "TRANSACTION_AMOUNT": "NUMERIC",
    "ACCOUNT_BALANCE": "NUMERIC",
    "CREDIT_SCORE": "NUMERIC",
    "INCOME": "NUMERIC",

    "TRANSACTION_HISTORY": "RECORD"
}

summary_analysis = summary.reset_index()

summary_analysis["field_type"] = (
    summary_analysis["field"]
    .map(field_type_map)
)

display(
    summary_analysis[
        [
            "field",
            "field_type",
            "TOTAL",
            "PARTIAL_VALUE",
            "EXACT_VALUE",
            "VALUE_REQUIREMENT_RATIO"
        ]
    ]
)
```

```python
# 14. 분석결과 기반 1차 Treatment 후보

def suggest_treatment(row):

    field_type = row["field_type"]
    exact = row["EXACT_REQUIREMENT_RATIO"]
    value = row["VALUE_REQUIREMENT_RATIO"]

    if field_type == "IDENTIFIER":

        if exact < 1:
            return "MASK / HMAC / TOKEN 후보"
        else:
            return "조건부 원문 필요 검토"

    elif field_type == "NUMERIC":

        if exact >= 10:
            return "KEEP_EXACT 필요성 높음"
        elif exact > 0:
            return "조건부 KEEP_EXACT"
        else:
            return "최소화 가능"

    elif field_type == "RECORD":
        return "별도 상세수준 분석 필요"

    return "REVIEW"


summary_analysis["treatment_candidate"] = (
    summary_analysis.apply(
        suggest_treatment,
        axis=1
    )
)

display(
    summary_analysis[
        [
            "field",
            "field_type",
            "VALUE_REQUIREMENT_RATIO",
            "EXACT_REQUIREMENT_RATIO",
            "treatment_candidate"
        ]
    ].sort_values(
        ["field_type", "VALUE_REQUIREMENT_RATIO"],
        ascending=[True, False]
    )
)
```

## 17. 가설검정 ① 식별정보와 금융 수치정보의 정확값 요구 차이

### 연구가설

- H0: 식별정보(IDENTIFIER)와 금융 수치정보(NUMERIC)의 EXACT_VALUE 요구 비율은 동일하다.
- H1: 식별정보의 EXACT_VALUE 요구 비율이 금융 수치정보보다 낮다.

### 분석 목적

상담 업무에서 데이터 유형에 따라 정확 원문값의 필요성이 실제로 다른지 검증한다.

유의한 차이가 확인될 경우,
모든 Field를 동일한 방식으로 처리하는 것이 아니라
데이터 유형별로 Transform 수준을 차등 설계해야 한다는 근거로 사용한다.

```python
# 17. 식별정보 vs 금융 수치정보 정확값 요구 비율 검정

from statsmodels.stats.proportion import proportions_ztest
from scipy.stats import chi2_contingency
import numpy as np

identifier = summary_analysis[
    summary_analysis["field_type"] == "IDENTIFIER"
]

numeric = summary_analysis[
    summary_analysis["field_type"] == "NUMERIC"
]

identifier_exact = identifier["EXACT_VALUE"].sum()
identifier_total = identifier["TOTAL"].sum()

numeric_exact = numeric["EXACT_VALUE"].sum()
numeric_total = numeric["TOTAL"].sum()

identifier_ratio = identifier_exact / identifier_total
numeric_ratio = numeric_exact / numeric_total

print("IDENTIFIER")
print(f"EXACT: {identifier_exact:,} / {identifier_total:,}")
print(f"비율: {identifier_ratio * 100:.3f}%")

print("\nNUMERIC")
print(f"EXACT: {numeric_exact:,} / {numeric_total:,}")
print(f"비율: {numeric_ratio * 100:.3f}%")
```

```python
# 18. 단측 두 비율 검정

count = np.array([
    identifier_exact,
    numeric_exact
])

nobs = np.array([
    identifier_total,
    numeric_total
])

z_stat, p_value = proportions_ztest(
    count,
    nobs,
    alternative="smaller"
)

print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.6g}")

if p_value < 0.05:
    print(
        "\n결론: H0 기각\n"
        "식별정보의 정확값 요구 비율은 금융 수치정보보다 "
        "통계적으로 유의하게 낮다."
    )
else:
    print(
        "\n결론: H0 기각 불가\n"
        "식별정보와 금융 수치정보 간 정확값 요구 비율의 "
        "유의한 차이를 확인하지 못했다."
    )
```

```python
# 19. 데이터 유형과 EXACT 요구 여부의 연관성 및 효과크기

contingency = np.array([
    [
        identifier_exact,
        identifier_total - identifier_exact
    ],
    [
        numeric_exact,
        numeric_total - numeric_exact
    ]
])

chi2, p, dof, expected = chi2_contingency(contingency)

n = contingency.sum()
phi = np.sqrt(chi2 / n)

print("Contingency Table")
print(contingency)

print(f"\nChi-square: {chi2:.4f}")
print(f"P-value: {p:.6g}")
print(f"Phi effect size: {phi:.4f}")
```

## 18. 가설검정 ② 금융 수치 Field별 정확값 요구 차이

### 연구가설

- H0: 금융 수치 Field별 EXACT_VALUE 요구 비율은 동일하다.
- H1: 금융 수치 Field별 EXACT_VALUE 요구 비율에는 차이가 있다.

### 분석 목적

금융 수치정보라고 하더라도
금리·잔액·거래금액·대출금액 등의 업무적 Utility가 동일한지 검증한다.

차이가 존재한다면 NUMERIC Field 역시 일괄 KEEP_EXACT이 아니라
Field별 필요 수준에 따라 처리해야 한다.

```python
# 20. 금융 수치 Field별 EXACT_VALUE 비율 차이 검정

numeric_test = summary_analysis[
    summary_analysis["field_type"] == "NUMERIC"
].copy()

numeric_test["NOT_EXACT"] = (
    numeric_test["TOTAL"]
    - numeric_test["EXACT_VALUE"]
)

contingency_numeric = numeric_test[
    [
        "EXACT_VALUE",
        "NOT_EXACT"
    ]
].to_numpy()

chi2, p, dof, expected = chi2_contingency(
    contingency_numeric
)

n = contingency_numeric.sum()

cramers_v = np.sqrt(
    chi2
    / (
        n
        * min(
            contingency_numeric.shape[0] - 1,
            contingency_numeric.shape[1] - 1
        )
    )
)

print(f"Chi-square: {chi2:.4f}")
print(f"df: {dof}")
print(f"P-value: {p:.6g}")
print(f"Cramer's V: {cramers_v:.4f}")

if p < 0.05:
    print(
        "\n결론: H0 기각\n"
        "금융 수치 Field별 EXACT_VALUE 요구 비율에는 "
        "통계적으로 유의한 차이가 있다."
    )
else:
    print(
        "\n결론: H0 기각 불가"
    )
```

## 19. 상담 단위 가설검정

### 가설

- H0: 동일 상담 내에서 식별정보와 금융 수치정보의 EXACT_VALUE 요구 확률에는 차이가 없다.
- H1: 식별정보보다 금융 수치정보에서 EXACT_VALUE가 더 자주 요구된다.

동일한 상담 안에서 두 데이터 유형을 비교하므로
상담 단위의 paired test인 McNemar 검정을 사용한다.

```python
# 21. 상담별 EXACT_VALUE 요구 여부 생성

identifier_fields = {
    "CUSTOMER_NAME",
    "RESIDENT_ID",
    "DATE_OF_BIRTH",
    "PHONE_NUMBER",
    "ACCOUNT_NUMBER",
    "CARD_NUMBER"
}

numeric_fields = {
    "INTEREST_RATE",
    "LOAN_AMOUNT",
    "DELINQUENCY_AMOUNT",
    "DELINQUENCY_PERIOD",
    "TRANSACTION_AMOUNT",
    "ACCOUNT_BALANCE",
    "CREDIT_SCORE",
    "INCOME"
}


def has_exact_requirement(requirements, field_set):

    return int(
        any(
            item["field"] in field_set
            and item["level"] == "EXACT_VALUE"
            for item in requirements
        )
    )


df_clean["identifier_exact"] = (
    df_clean["data_requirements_v2"]
    .apply(
        lambda x: has_exact_requirement(
            x,
            identifier_fields
        )
    )
)

df_clean["numeric_exact"] = (
    df_clean["data_requirements_v2"]
    .apply(
        lambda x: has_exact_requirement(
            x,
            numeric_fields
        )
    )
)

print(
    "식별정보 EXACT 상담:",
    df_clean["identifier_exact"].sum()
)

print(
    "금융수치 EXACT 상담:",
    df_clean["numeric_exact"].sum()
)
```

```python
# 22. McNemar 검정

from statsmodels.stats.contingency_tables import mcnemar
import pandas as pd

table = pd.crosstab(
    df_clean["identifier_exact"],
    df_clean["numeric_exact"]
)

table = table.reindex(
    index=[0, 1],
    columns=[0, 1],
    fill_value=0
)

display(table)

result = mcnemar(
    table,
    exact=True
)

print(f"McNemar statistic: {result.statistic}")
print(f"P-value: {result.pvalue:.6g}")

if result.pvalue < 0.05:
    print(
        "\n결론: H0 기각\n"
        "동일 상담 내에서도 금융 수치정보의 정확값 요구가 "
        "식별정보보다 유의하게 높다."
    )
else:
    print("\n결론: H0 기각 불가")
```

## 종합 결론

세 가지 검정은 동일한 방향을 지지한다.

> **금융 상담에서 요구되는 데이터 Utility는 데이터 유형 및 Field별로 유의하게 다르다.**

따라서 Gateway의 Transform 정책은
모든 정보를 동일하게 마스킹하거나 동일하게 원문 유지하는 방식이 아니라,

**Field → 업무상 필요 수준 → Utility Requirement → Transform Treatment**

순으로 결정되어야 한다.

특히 상담 데이터에서는 식별정보의 정확 원문값 필요성이 매우 낮게 관측된 반면,
금리·연체기간 등 일부 금융 수치정보에서는 정확값 보존 필요성이 상대적으로 높게 나타났다.

따라서 다음 단계에서는 이 결과를 기반으로
각 Field를 `식별 유지 필요 / 관계 유지 필요 / 복원 필요 / 정확값 필요`로 구분한 뒤
MASK / HMAC / TOKEN / KEEP_EXACT Treatment와 연결한다.

## 20. Field별 Utility Requirement Matrix

상담 데이터 분석 결과를 기반으로 각 Field에 대해
Transform 기법을 바로 결정하지 않고 다음 업무 요구조건을 정의한다.

- Exact Requirement: 정확한 원문값이 필요한가
- Partial Requirement: 부분값만으로 업무가 가능한가
- Relationship Requirement: 동일 고객·계좌·거래 간 관계 유지가 필요한가
- Reversibility Requirement: 이후 원본 복원이 필요한가

이 4개 조건을 기반으로 최종 Transform Treatment를 결정한다.
