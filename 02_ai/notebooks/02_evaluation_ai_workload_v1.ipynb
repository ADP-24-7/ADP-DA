## 1. 상담 대화 Turn 분리

### 분석 목적

원본 `consulting_content`에는 상담원과 고객의 발화가 하나의 문자열 안에 함께 저장되어 있다.

본 분석에서는 원본 데이터의 `TX`, `RX` 표기를 기준으로 상담 대화를 발화 단위로 분리한다.

- `TX` : 상담원 발화
- `RX` : 고객 발화

각 상담 건은 `source_id`를 기준으로 유지하고, 별도의 `turn_df`에서는 하나의 발화를 하나의 행으로 구성한다.

### 분석 이유

이후 Intent 분석에서는 고객이 실제로 어떤 업무를 요청했는지를 확인해야 하므로, 고객 발화와 상담원 발화를 분리할 필요가 있다.

또한 발화 단위로 분리하면 다음 분석이 가능하다.

- 고객 요청 Intent 분석
- 상담원 응답 패턴 분석
- 상담 주제별 발화 구조 비교
- 개인정보 및 주요 데이터 필드 등장 위치 분석
- 향후 Transform 전후 상담 품질 비교

### 처리 방식

`consulting_content` 내부의 `TX`, `RX` 위치를 기준으로 각 발화를 분리하고 다음 형태로 재구성한다.

| 컬럼 | 의미 |
|---|---|
| source_id | 원본 상담 식별자 |
| consulting_topic | 상담 주제 |
| turn_index | 상담 내 발화 순서 |
| raw_role | 원본 TX / RX 값 |
| speaker | 상담원 / 고객 |
| utterance | 실제 발화 내용 |

분리 후에는 전체 상담 건 중 Turn 분리가 실패한 데이터가 존재하는지 추가로 검증한다.

```python
import re

import pandas as pd

# 06. 상담 대화 Turn 분리
# 정제 완료된 df_clean의 consulting_content를 기준으로 TX/RX 발화를 분리한다.
# 원본 상담 1건은 유지하면서, 별도의 turn_df에서는 1행을 하나의 발화 Turn으로 구성한다.

pattern = re.compile(r'(?:^|\n)\s*(TX|RX)\s*')

def parse_turns(text):
    text = str(text)
    matches = list(pattern.finditer(text))
    
    turns = []

    for i, match in enumerate(matches):
        role = match.group(1)

        start = match.end()
        end = matches[i + 1].start() if i + 1 < len(matches) else len(text)

        utterance = text[start:end].strip()

        if utterance:
            turns.append({
                "raw_role": role,
                "speaker": "상담원" if role == "TX" else "고객",
                "utterance": utterance
            })

    return turns


# df_clean 기준으로 Turn 분리
turn_rows = []

for _, row in df_clean.iterrows():

    turns = parse_turns(row["consulting_content"])

    for turn_index, turn in enumerate(turns, start=1):

        turn_rows.append({
            "source_id": row["source_id"],
            "consulting_topic": row["consulting_topic"],
            "turn_index": turn_index,
            "raw_role": turn["raw_role"],
            "speaker": turn["speaker"],
            "utterance": turn["utterance"]
        })

turn_df = pd.DataFrame(turn_rows)

print("정제 상담 건수:", f"{len(df_clean):,}")
print("전체 발화 Turn:", f"{len(turn_df):,}")
```

```python
# 08. 상담내용 유형 분류

def classify_content(text):
    text = str(text)

    has_tx = "TX" in text
    has_rx = "RX" in text

    if has_tx or has_rx:
        return "dialogue"
    else:
        return "summary"


df["content_type"] = df["consulting_content"].apply(classify_content)

print(df["content_type"].value_counts())
print("\n비율")
print((df["content_type"].value_counts(normalize=True) * 100).round(2))
```

## 1. 상담내용 유형 분리

`consulting_content`를 확인한 결과, 상담 내용은 실제 대화형 데이터와 상담 요약형 데이터가 혼재되어 있었다.

### 1.1 Dialogue 유형

`TX`, `RX` 표기가 포함된 실제 상담 대화 형태이다.

- `TX` : 상담원 발화
- `RX` : 고객 발화
- 실제 발화 순서를 유지하고 있어 Turn 단위 분석이 가능하다.

### 1.2 Summary 유형

`TX`, `RX` 구분 없이 상담 전체 내용을 서술형으로 요약한 형태이다.

예를 들어 고객의 문의 내용과 상담원의 응답 내용이 하나의 문단 안에 함께 기록되어 있다.

따라서 Summary 데이터를 상담원 발화로 임의 변환하지 않고, 별도의 상담 요약 데이터로 유지한다.

### 1.3 유형별 데이터 분포

전체 26,563건의 상담 데이터를 분류한 결과 다음과 같이 나타났다.

| content_type | 건수 | 비율 |
|---|---:|---:|
| Summary | 19,427 | 73.14% |
| Dialogue | 7,136 | 26.86% |
| 전체 | 26,563 | 100.00% |

전체 데이터의 약 73%가 Summary 형태이고, 약 27%가 실제 Dialogue 형태로 구성되어 있다.

따라서 이후 분석에서는 하나의 방식으로 강제 변환하지 않고 두 데이터 유형의 구조적 차이를 유지한다.

- Dialogue 데이터는 고객 발화(RX)를 중심으로 Intent를 분석한다.
- Summary 데이터는 문장 내 고객의 문의 내용을 중심으로 Intent를 분석한다.

최종적으로 두 유형에서 고객이 실제로 요청한 내용을 공통 형태로 추출하여 동일한 Intent 분석 체계에 연결한다.

```python
# 20. 전체 상담 Topic 분포

topic_count = (
    df_clean["consulting_topic"]
    .value_counts()
    .reset_index()
)

topic_count.columns = ["consulting_topic", "count"]

topic_count["ratio"] = (
    topic_count["count"] / len(df_clean) * 100
).round(2)

print("전체 상담:", f"{len(df_clean):,}건")
print("Topic 수:", df_clean["consulting_topic"].nunique())

display(topic_count)
```

```python
# 22. Topic × 상담 형식 교차분석

topic_type = pd.crosstab(
    df_clean["consulting_topic"],
    df_clean["content_type"]
)

topic_type["total"] = topic_type.sum(axis=1)

if "dialogue" in topic_type.columns:
    topic_type["dialogue_ratio"] = (
        topic_type["dialogue"] / topic_type["total"] * 100
    ).round(2)

if "summary" in topic_type.columns:
    topic_type["summary_ratio"] = (
        topic_type["summary"] / topic_type["total"] * 100
    ).round(2)

display(
    topic_type.sort_values("total", ascending=False)
)
```

```python
# 23. Intent 분석용 텍스트 생성
# Dialogue와 Summary의 전체 상담 맥락은 유지한다.
# 단, Dialogue에만 존재하는 TX/RX 표기는 형식적 차이이므로 제거한다.

def normalize_analysis_text(text):
    text = str(text)

    # TX/RX 역할 표지만 제거
    text = re.sub(r'(^|\n)\s*(TX|RX)\s*', ' ', text)

    # 연속 공백 정리
    text = re.sub(r'\s+', ' ', text).strip()

    return text


df_clean["analysis_text"] = (
    df_clean["consulting_content"]
    .apply(normalize_analysis_text)
)

print("분석 대상:", f"{len(df_clean):,}")
print("빈 텍스트:", df_clean["analysis_text"].str.strip().eq("").sum())

display(
    df_clean[
        ["consulting_topic", "content_type", "analysis_text"]
    ].sample(5, random_state=42)
)
```

```python
from sentence_transformers import SentenceTransformer

# 24. Multilingual Sentence Embedding

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

embeddings = model.encode(
    df_clean["analysis_text"].tolist(),
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True
)

print("Embedding shape:", embeddings.shape)
```

## 4. 기존 9개 상담 Topic의 의미적 타당성 검증

원천 데이터에는 이미 9개의 `consulting_topic`이 업무 분류값으로 제공되어 있다.
따라서 기존 9개 업무 분류를 기준으로 실제 상담 텍스트의 의미적 분리가 얼마나 명확한지를 검증한다.

Sentence Embedding은 새로운 군집을 만들기 위한 용도가 아니라 다음을 확인하는 데 활용한다.

1. 동일 Topic의 상담내용이 의미적으로 서로 가까운지
2. 서로 다른 Topic 간 의미적 중첩이 어느 정도 존재하는지
3. 어떤 Topic끼리 실제 상담 맥락에서 혼동될 가능성이 높은지
4. 기존 업무분류 내부에 다양한 세부 업무가 얼마나 존재하는지

이를 통해 기존 9개 Topic을 FPG의 상위 Workload/Intent 분류 기준으로 사용할 수 있는지 검증한다.

이후 각 Topic별로 실제 상담 과정에서 요구되는 정보와 업무행위를 추출하여
`Topic → 업무행위 → 필요 데이터 → Transform 방식`으로 연결한다.

```python
# 25. Sentence Embedding 기반 기존 Topic 분류력 검증

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Topic 라벨 인코딩
le = LabelEncoder()

y = le.fit_transform(
    df_clean["consulting_topic"]
)

X_train, X_test, y_train, y_test = train_test_split(
    embeddings,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

clf = LogisticRegression(
    max_iter=3000,
    class_weight="balanced"
)

clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred), 4)
)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=le.classes_,
        digits=4
    )
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap
from sklearn.metrics import confusion_matrix

# 28. 네이비 계열 컬러맵
navy_cmap = LinearSegmentedColormap.from_list(
    "navy_blue",
    [
        "#EEF6F8",
        "#AED9DD",
        "#6AB1CC",
        "#2D78A6",
        "#163D66",
        "#0F2747"
    ]
)

# 혼동행렬
cm = confusion_matrix(
    y_test,
    y_pred,
    normalize="true"
)

fig, ax = plt.subplots(figsize=(11, 9))

im = ax.imshow(
    cm,
    cmap=navy_cmap,
    vmin=0,
    vmax=1
)

# 축 설정
ax.set_xticks(np.arange(len(le.classes_)))
ax.set_yticks(np.arange(len(le.classes_)))

ax.set_xticklabels(
    le.classes_,
    rotation=45,
    ha="right"
)

ax.set_yticklabels(le.classes_)

ax.set_xlabel("예측 Topic")
ax.set_ylabel("실제 Topic")

ax.set_title(
    "상담 Topic 의미적 분류 혼동행렬",
    fontsize=15,
    fontweight="bold"
)

# 셀 값 표시
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):

        value = cm[i, j]

        text_color = (
            "white"
            if value >= 0.45
            else "black"
        )

        ax.text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center",
            color=text_color,
            fontsize=9,
            fontweight="bold"
        )

plt.colorbar(
    im,
    ax=ax,
    fraction=0.046,
    pad=0.04
)

plt.tight_layout()
plt.show()
```

```python
# 27. Topic별 Embedding 중심점 계산

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

topic_centroids = []
topic_names = []

for topic in df_clean["consulting_topic"].unique():

    idx = (
        df_clean["consulting_topic"] == topic
    ).values

    centroid = embeddings[idx].mean(axis=0)

    topic_centroids.append(centroid)
    topic_names.append(topic)

topic_centroids = np.vstack(topic_centroids)

similarity_matrix = cosine_similarity(
    topic_centroids
)

topic_similarity = pd.DataFrame(
    similarity_matrix,
    index=topic_names,
    columns=topic_names
)

display(topic_similarity.round(3))
```

```python
# 29. Topic 간 의미적 유사도 Heatmap

similarity_values = topic_similarity.values

fig, ax = plt.subplots(figsize=(12, 10))

im = ax.imshow(
    similarity_values,
    cmap=navy_cmap,
    vmin=similarity_values.min(),
    vmax=1
)

ax.set_xticks(
    np.arange(len(topic_similarity.columns))
)

ax.set_yticks(
    np.arange(len(topic_similarity.index))
)

ax.set_xticklabels(
    topic_similarity.columns,
    rotation=45,
    ha="right"
)

ax.set_yticklabels(
    topic_similarity.index
)

ax.set_title(
    "상담 Topic 간 의미적 유사도",
    fontsize=15,
    fontweight="bold"
)

# 셀 수치 표시
threshold = (
    similarity_values.min()
    + similarity_values.max()
) / 2

for i in range(similarity_values.shape[0]):
    for j in range(similarity_values.shape[1]):

        value = similarity_values[i, j]

        text_color = (
            "white"
            if value >= threshold
            else "black"
        )

        ax.text(
            j,
            i,
            f"{value:.3f}",
            ha="center",
            va="center",
            color=text_color,
            fontsize=9,
            fontweight="bold"
        )

plt.colorbar(
    im,
    ax=ax,
    fraction=0.046,
    pad=0.04
)

plt.tight_layout()
plt.show()
```

따라서 이후 분석에서는 Topic 자체보다 실제 상담에서 수행되는 세부 업무행위와 필요한 데이터 필드를 중심으로 분석한다.

**Topic → 업무행위(Action) → 필요 데이터(Field) → Transform 방식**

이 구조를 통해 동일한 Topic 내부에서도 실제 업무 목적에 따라 서로 다른 데이터 처리 정책을 적용할 수 있도록 확장한다.

## 5. Topic별 실제 업무행위(Action) 추출

앞선 분석에서 기존 9개 Topic은 상위 업무 분류로 유효하지만,
실제 상담에서는 여러 업무 요소가 동시에 나타나는 것을 확인하였다.

따라서 각 상담의 전체 `consulting_content`를 기준으로 실제 수행·요청된 업무행위를 다중 라벨로 추출한다.

하나의 상담에는 `조회 + 변경`, `연장 + 신청`, `이체 + 확인`처럼 여러 Action이 동시에 존재할 수 있으므로
하나의 상담을 하나의 Action으로 강제 분류하지 않는다.

이를 통해 각 Topic에서 어떤 업무행위가 실제로 반복되는지 확인하고,
이후 해당 Action에 필요한 데이터 Field 분석으로 연결한다.

```python
# 30. 금융 상담 업무행위(Action) 사전 정의

ACTION_PATTERNS = {
    "조회": r"조회|내역.*보|현황.*보|잔액.*보",
    "확인": r"확인|확인하|확인해",
    "신청": r"신청|신청하|접수",
    "변경": r"변경|바꾸|수정",
    "해지": r"해지|해약",
    "연장": r"연장|갱신",
    "이체": r"이체|자동이체",
    "송금": r"송금",
    "입금": r"입금",
    "출금": r"출금",
    "상환": r"상환|중도상환|원금.*갚|대출.*갚",
    "납부": r"납부",
    "환전": r"환전|외화.*교환|원화.*전환",
    "한도조정": r"한도.*변경|한도.*상향|한도.*하향|한도.*해제",
    "정지": r"정지|거래중지|지급정지",
    "정지해제": r"정지.*해제|거래중지.*해제|지급정지.*해제",
    "재발급": r"재발급|다시.*발급",
    "인증": r"인증|OTP|본인확인|본인 인증",
    "신고": r"신고|사기.*접수|착오송금.*접수",
    "취소": r"취소",
    "환불": r"환불",
    "계산": r"계산|산정",
    "안내": r"안내|설명",
}
```

```python
# 31. 상담별 Action 추출

def extract_actions(text):
    text = str(text)

    detected = []

    for action, pattern in ACTION_PATTERNS.items():
        if re.search(pattern, text):
            detected.append(action)

    return detected


df_clean["actions"] = (
    df_clean["analysis_text"]
    .apply(extract_actions)
)

df_clean["action_count"] = (
    df_clean["actions"]
    .apply(len)
)

print("전체 상담:", f"{len(df_clean):,}건")
print(
    "Action 1개 이상 탐지:",
    f"{(df_clean['action_count'] > 0).sum():,}건"
)
print(
    "Action 미탐지:",
    f"{(df_clean['action_count'] == 0).sum():,}건"
)

display(
    df_clean[
        [
            "consulting_topic",
            "analysis_text",
            "actions",
            "action_count"
        ]
    ].sample(10, random_state=42)
)
```

```python
# 32. 상담 × Action 구조 변환

action_long = (
    df_clean[
        ["source_id", "consulting_topic", "actions"]
    ]
    .explode("actions")
    .dropna(subset=["actions"])
    .rename(columns={"actions": "action"})
)

print("Action 발생 건수:", f"{len(action_long):,}")
print("Action 종류:", action_long["action"].nunique())
```

```python
# 33. Action별 상담 건수 및 침투율

action_summary = (
    action_long
    .groupby("action")["source_id"]
    .nunique()
    .reset_index(name="consult_count")
)

action_summary["consult_ratio"] = (
    action_summary["consult_count"]
    / len(df_clean)
    * 100
).round(2)

action_summary = (
    action_summary
    .sort_values(
        "consult_count",
        ascending=False
    )
    .reset_index(drop=True)
)

display(action_summary)
```

### 5.1 Topic별 Action 기술통계 및 식별력 분석

앞선 분석에서 9개 상담 Topic은 서로 완전히 독립된 영역이라기보다,
실제 상담 과정에서 여러 업무행위(Action)를 공유하는 구조를 가진다는 점을 확인하였다.

이에 따라 각 Topic에서 어떤 Action이 실제로 얼마나 자주 등장하는지 먼저 기술통계로 확인한다.

하나의 상담에는 조회·이체·해지·연장·입금·상환 등 여러 Action이 동시에 나타날 수 있으므로,
상담 1건을 하나의 Action으로 강제 분류하지 않고 다중 Action 구조로 분석한다.

#### 1. Action 등장률

각 Topic에서 Action이 포함된 상담 건수와 비율을 계산하여
실제 고객상담 과정에서 수행되는 업무행위의 분포를 확인한다.

그러나 단순 등장률만으로 특정 Topic의 특징적인 Action이라고 판단할 수는 없다.

예를 들어 `확인`, `안내`, `신청`과 같은 표현은 여러 금융업무에서 공통적으로 나타날 수 있기 때문이다.

#### 2. Action Lift

따라서 특정 Topic에서 Action이 전체 상담보다 상대적으로 얼마나 집중되어 있는지를 확인하기 위해
Action Lift를 계산한다.

**Lift = 특정 Topic 내 Action 등장률 / 전체 상담 내 Action 등장률**

- Lift > 1 : 해당 Topic에서 상대적으로 많이 등장
- Lift ≈ 1 : 금융 상담 전반에서 공통적으로 등장
- Lift < 1 : 해당 Topic에서는 상대적으로 덜 등장

이를 통해 단순히 자주 등장하는 범용 Action과
특정 금융업무를 상대적으로 특징짓는 Action을 구분한다.

다만 Lift는 기술통계적 상대비율이므로,
해당 차이가 우연에 의해 발생했는지는 판단할 수 없다.

따라서 다음 단계에서는 Topic과 Action 간 관계에 대해 통계적 가설검정을 수행한다.

```python
# 37. Action 분석용 Long Format

action_long = (
    df_clean[
        ["source_id", "consulting_topic", "actions"]
    ]
    .explode("actions")
    .dropna(subset=["actions"])
    .rename(columns={"actions": "action"})
)

print("전체 상담:", f"{len(df_clean):,}건")
print("Action 종류:", action_long["action"].nunique())
print("Action 발생 건수:", f"{len(action_long):,}")
```

### 5.2 Topic과 Action의 전체 연관성 가설검정

기술통계와 Lift 분석을 통해 Topic별 Action 구성에 차이가 있음을 탐색적으로 확인하였다.

이 차이가 우연한 표본 변동인지 검증하기 위해 각 Action에 대해
`9개 Topic × Action 발생 여부` 교차표를 구성하고 카이제곱 독립성 검정을 수행한다.

**H0 (귀무가설)**  
상담 Topic과 해당 Action의 발생 여부는 서로 독립이다.

**H1 (대립가설)**  
상담 Topic에 따라 해당 Action의 발생 비율이 다르다.

표본 수가 크기 때문에 p-value뿐 아니라 연관성의 크기를 나타내는 Cramér's V를 함께 사용한다.
또한 여러 Action을 동시에 검정하므로 Benjamini-Hochberg 방식으로 FDR을 보정한다.

```python
# 40. Action별 전체 가설검정

import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests

chi_rows = []

actions = sorted(action_long["action"].unique())

for action in actions:

    action_ids = set(
        action_long.loc[
            action_long["action"] == action,
            "source_id"
        ]
    )

    temp = df_clean[
        ["source_id", "consulting_topic"]
    ].copy()

    temp["action_present"] = (
        temp["source_id"]
        .isin(action_ids)
        .astype(int)
    )

    table = pd.crosstab(
        temp["consulting_topic"],
        temp["action_present"]
    )

    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.to_numpy().sum()
    r, c = table.shape

    cramers_v = np.sqrt(
        chi2 /
        (n * min(r - 1, c - 1))
    )

    chi_rows.append({
        "action": action,
        "chi2": chi2,
        "dof": dof,
        "p_value": p_value,
        "cramers_v": cramers_v,
        "min_expected": expected.min()
    })


action_chi2 = pd.DataFrame(chi_rows)

action_chi2["p_adj"] = multipletests(
    action_chi2["p_value"],
    method="fdr_bh"
)[1]

action_chi2["significant"] = (
    action_chi2["p_adj"] < 0.05
)

display(
    action_chi2
    .sort_values(
        "cramers_v",
        ascending=False
    )
    .round(4)
)
```

```python
# 41. 효과크기 해석

def cramers_strength(v):

    if v >= 0.5:
        return "LARGE"
    elif v >= 0.3:
        return "MODERATE"
    elif v >= 0.1:
        return "SMALL"
    else:
        return "VERY_SMALL"


action_chi2["effect_size"] = (
    action_chi2["cramers_v"]
    .apply(cramers_strength)
)

display(
    action_chi2[
        [
            "action",
            "chi2",
            "p_adj",
            "cramers_v",
            "effect_size",
            "significant"
        ]
    ]
    .sort_values(
        "cramers_v",
        ascending=False
    )
    .round(4)
)
```

```python
# 42. Action별 Cramér's V 시각화

import matplotlib.pyplot as plt

plot_df = (
    action_chi2
    .sort_values(
        "cramers_v",
        ascending=False
    )
    .copy()
)

navy_palette = [
    "#0F2747",
    "#163D66",
    "#1F5A85",
    "#2D78A6",
    "#4A97BD",
    "#6AB1CC",
    "#8BC6D4",
    "#AED9DD",
    "#D2E9E8"
]

colors = [
    navy_palette[
        min(
            int(i * len(navy_palette) / len(plot_df)),
            len(navy_palette) - 1
        )
    ]
    for i in range(len(plot_df))
]

plt.figure(figsize=(11, 8))

bars = plt.barh(
    plot_df["action"],
    plot_df["cramers_v"],
    color=colors
)

plt.gca().invert_yaxis()

plt.xlabel("Cramér's V")
plt.ylabel("Action")

plt.title(
    "상담 Topic과 업무행위(Action)의 연관성 효과크기",
    fontsize=15,
    fontweight="bold"
)

for bar, value in zip(
    bars,
    plot_df["cramers_v"],
    strict=False,
):
    plt.text(
        value + 0.005,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.3f}",
        va="center"
    )

plt.tight_layout()
plt.show()
```

### 5.3 Topic-Action 사후검정

전체 가설검정에서 Topic에 따라 Action 발생구조가 달라지는지를 검증한 뒤,
구체적으로 어떤 Topic-Action 조합이 이러한 차이를 만드는지 사후검정을 수행한다.

각 Topic을 나머지 Topic 전체와 비교하는 2×2 분할표를 구성하고
Fisher's Exact Test를 수행한다.

관계의 방향과 크기를 함께 평가하기 위해 다음 지표를 사용한다.

- Lift
- Odds Ratio
- Odds Ratio 95% 신뢰구간
- FDR 보정 p-value

특정 Topic에서 해당 Action이 상대적으로 집중된 관계는
Lift > 1, Odds Ratio > 1, 신뢰구간 하한 > 1을 기준으로 확인한다.

```python
# 43. Topic-Action 사후검정

from scipy.stats import fisher_exact

posthoc_rows = []

topics = df_clean["consulting_topic"].unique()

all_ids = set(df_clean["source_id"])

for topic in topics:

    topic_ids = set(
        df_clean.loc[
            df_clean["consulting_topic"] == topic,
            "source_id"
        ]
    )

    other_ids = all_ids - topic_ids

    for action in actions:

        action_ids = set(
            action_long.loc[
                action_long["action"] == action,
                "source_id"
            ]
        )

        a = len(topic_ids & action_ids)
        b = len(topic_ids - action_ids)
        c = len(other_ids & action_ids)
        d = len(other_ids - action_ids)

        fisher_or, p_value = fisher_exact(
            [[a, b], [c, d]],
            alternative="two-sided"
        )

        # 0 cell 대응
        aa = a + 0.5
        bb = b + 0.5
        cc = c + 0.5
        dd = d + 0.5

        corrected_or = (
            aa * dd
        ) / (
            bb * cc
        )

        log_or = np.log(corrected_or)

        se = np.sqrt(
            1/aa +
            1/bb +
            1/cc +
            1/dd
        )

        ci_low = np.exp(
            log_or - 1.96 * se
        )

        ci_high = np.exp(
            log_or + 1.96 * se
        )

        posthoc_rows.append({
            "consulting_topic": topic,
            "action": action,
            "count": a,
            "odds_ratio": fisher_or,
            "or_ci_low": ci_low,
            "or_ci_high": ci_high,
            "p_value": p_value
        })


action_posthoc = pd.DataFrame(posthoc_rows)

action_posthoc["p_adj"] = multipletests(
    action_posthoc["p_value"],
    method="fdr_bh"
)[1]
```

```python
# 44. 기술통계 + 가설검정 결과 결합

action_evidence = action_lift.merge(
    action_posthoc[
        [
            "consulting_topic",
            "action",
            "odds_ratio",
            "or_ci_low",
            "or_ci_high",
            "p_adj"
        ]
    ],
    on=["consulting_topic", "action"],
    how="left"
)

display(
    action_evidence[
        [
            "consulting_topic",
            "action",
            "count",
            "ratio",
            "lift",
            "odds_ratio",
            "or_ci_low",
            "or_ci_high",
            "p_adj"
        ]
    ]
    .sort_values(
        ["consulting_topic", "lift"],
        ascending=[True, False]
    )
    .round(3)
)
```

```python
display(
    action_chi2[
        [
            "action",
            "chi2",
            "dof",
            "p_adj",
            "cramers_v",
            "effect_size",
            "significant"
        ]
    ]
    .sort_values(
        "cramers_v",
        ascending=False
    )
    .round(4)
)
```

```python
# 45. 유의하고 방향성이 안정적인 Topic-Action 관계

core_action = action_evidence[
    (action_evidence["count"] >= 20)
    & (action_evidence["lift"] > 1)
    & (action_evidence["odds_ratio"] > 1)
    & (action_evidence["or_ci_low"] > 1)
    & (action_evidence["p_adj"] < 0.05)
].copy()


def classify_strength(lift):

    if lift >= 2:
        return "STRONG"

    elif lift >= 1.5:
        return "MODERATE"

    return "WEAK"


core_action["strength"] = (
    core_action["lift"]
    .apply(classify_strength)
)

display(
    core_action[
        [
            "consulting_topic",
            "action",
            "count",
            "ratio",
            "lift",
            "odds_ratio",
            "or_ci_low",
            "or_ci_high",
            "p_adj",
            "strength"
        ]
    ]
    .sort_values(
        ["consulting_topic", "lift"],
        ascending=[True, False]
    )
    .round(3)
)
```

```python
# 46. Topic별 실질적으로 사용되는 Action 정의
# Topic 상담의 5% 이상에서 등장한 Action을 사용

active_action = topic_action_count[
    topic_action_count["ratio"] >= 5
].copy()

topic_action_sets = (
    active_action
    .groupby("consulting_topic")["action"]
    .apply(set)
)

for topic, action_set in topic_action_sets.items():

    print(f"\n[{topic}]")
    print(sorted(action_set))
```

```python
# 47. Topic 간 Action 공유도 계산

topic_names = topic_action_sets.index.tolist()

jaccard_matrix = pd.DataFrame(
    index=topic_names,
    columns=topic_names,
    dtype=float
)

for topic_a in topic_names:

    set_a = topic_action_sets[topic_a]

    for topic_b in topic_names:

        set_b = topic_action_sets[topic_b]

        union = set_a | set_b
        intersection = set_a & set_b

        jaccard = (
            len(intersection) / len(union)
            if len(union) > 0
            else 0
        )

        jaccard_matrix.loc[
            topic_a,
            topic_b
        ] = jaccard


display(
    jaccard_matrix.round(3)
)
```

```python
# 47. Topic 간 Action 공유도 계산

topic_names = topic_action_sets.index.tolist()

jaccard_matrix = pd.DataFrame(
    index=topic_names,
    columns=topic_names,
    dtype=float
)

for topic_a in topic_names:

    set_a = topic_action_sets[topic_a]

    for topic_b in topic_names:

        set_b = topic_action_sets[topic_b]

        union = set_a | set_b
        intersection = set_a & set_b

        jaccard = (
            len(intersection) / len(union)
            if len(union) > 0
            else 0
        )

        jaccard_matrix.loc[
            topic_a,
            topic_b
        ] = jaccard


display(
    jaccard_matrix.round(3)
)
```

```python
# 49. Topic 간 공유도 순위

pair_rows = []

for i in range(len(topic_names)):

    for j in range(i + 1, len(topic_names)):

        topic_a = topic_names[i]
        topic_b = topic_names[j]

        shared = (
            topic_action_sets[topic_a]
            & topic_action_sets[topic_b]
        )

        pair_rows.append({
            "topic_a": topic_a,
            "topic_b": topic_b,
            "jaccard": jaccard_matrix.loc[
                topic_a,
                topic_b
            ],
            "shared_action_count": len(shared),
            "shared_actions": ", ".join(sorted(shared))
        })


topic_overlap = (
    pd.DataFrame(pair_rows)
    .sort_values(
        "jaccard",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    topic_overlap.round(3)
)
```

### 5.2 Topic과 Action의 통계적 연관성 검정

#### 가설 설정

9개 상담 Topic에 따라 실제 상담에서 수행되는 업무행위(Action)의 발생 구조가 달라지는지 검정하였다.

**H0 (귀무가설)**  
상담 Topic과 업무행위(Action)의 발생 여부는 서로 독립이다.

**H1 (대립가설)**  
상담 Topic에 따라 업무행위(Action)의 발생 비율이 다르다.

각 Action에 대해 `9개 Topic × Action 발생 여부` 교차표를 구성하고 카이제곱 독립성 검정을 수행하였다.

여러 Action을 동시에 검정하기 때문에 Benjamini-Hochberg 방식으로 p-value를 보정하였으며,
표본 크기가 큰 데이터에서 통계적 유의성이 과대평가되는 것을 방지하기 위해
Cramér's V를 함께 산출하여 연관성의 효과크기를 평가하였다.


#### 검정 결과

분석한 모든 Action에서 FDR 보정 후에도 통계적으로 유의한 결과가 확인되었다.

※ 출력상 `p_adj = 0.0000`은 실제 p-value가 정확히 0이라는 의미가 아니라,
표시 자릿수보다 작은 매우 작은 값이라는 의미이다.

효과크기가 특히 크게 나타난 Action은 다음과 같다.

| Action | χ² | Cramér's V | 효과크기 |
|---|---:|---:|---|
| 환전 | 21,962.02 | 0.909 | LARGE |
| 연장 | 21,733.88 | 0.905 | LARGE |
| 한도조정 | 14,767.18 | 0.746 | LARGE |
| 해지 | 10,852.08 | 0.639 | LARGE |
| 이체 | 8,372.69 | 0.561 | LARGE |
| 입금 | 6,737.28 | 0.504 | LARGE |
| 상환 | 6,133.23 | 0.481 | MODERATE |
| 송금 | 5,639.55 | 0.461 | MODERATE |
| 납부 | 4,741.67 | 0.423 | MODERATE |
| 신청 | 3,210.06 | 0.348 | MODERATE |

반면 `안내`, `재발급`, `정지해제`는 통계적으로 유의하더라도
Cramér's V가 0.1 미만으로 나타나 Topic을 구별하는 효과는 매우 작았다.

따라서 단순히 p-value가 유의하다는 이유만으로 해당 Action을
Topic의 핵심 업무행위라고 판단하지 않고 효과크기를 함께 고려하였다.


#### 가설 판단

전체 Action에서 보정 p-value가 유의수준 0.05보다 작게 나타났으므로
**상담 Topic과 Action이 서로 독립이라는 귀무가설 H0를 기각한다.**

즉, 고객이 어떤 상담 Topic으로 상담하느냐에 따라
실제로 나타나는 업무행위의 발생 비율은 통계적으로 동일하지 않다.

특히 `환전`, `연장`, `한도조정`, `해지`, `이체`, `입금`은
Topic에 따른 차이가 매우 큰 Action으로 나타났다.

이는 기존 9개 Topic이 단순한 명칭상의 분류가 아니라
실제 상담 과정에서도 서로 다른 업무행위의 강도와 조합을 가지는 업무영역임을 의미한다.


### 5.3 Topic 간 Action 공유 구조

그러나 Topic별 Action 구성비가 다르다는 사실이
각 금융업무가 서로 완전히 독립되어 있다는 것을 의미하지는 않는다.

Topic별로 일정 수준 이상 등장한 Action 집합의 Jaccard Similarity를 계산한 결과,
다수의 Topic 조합에서 높은 Action 공유도가 확인되었다.

대표적으로 다음과 같다.

| Topic 조합 | Jaccard Similarity | 공유 Action 수 |
|---|---:|---:|
| 부수거래금리감면 ↔ 이자/연체금액 | 0.929 | 13 |
| 대출문의 ↔ 이자/연체금액 | 0.923 | 12 |
| 거래내역/잔액조회 ↔ 중계요청/착오송금 | 0.867 | 13 |
| 대출문의 ↔ 부수거래금리감면 | 0.857 | 12 |
| 만기·연장/해지·수신 ↔ 부수거래금리감면 | 0.857 | 12 |
| 대출문의 ↔ 만기·연장/해지·수신 | 0.846 | 11 |

즉, 각 Topic에서 특정 Action이 나타나는 **빈도와 강도는 통계적으로 다르지만**,
조회·변경·신청·이체·인증·입금·출금·확인 등 실제 상담에서 사용되는
업무행위 자체는 여러 Topic에 걸쳐 상당 부분 공유되고 있었다.


### 5.4 최종 분석 결론

본 분석에서는 두 가지 사실이 동시에 확인되었다.

첫째, 카이제곱 검정과 Cramér's V를 통해
**9개 상담 Topic은 서로 다른 Action 구성과 강도를 가진다.**

둘째, Jaccard Similarity를 통해
**서로 다른 Topic이라도 실제 상담에서 사용하는 다수의 Action은 공유된다.**

따라서 실제 은행 고객상담은
`예금 업무`, `대출 업무`, `이체 업무`처럼 완전히 독립된 업무 단위로만 작동한다고 보기 어렵다.

고객의 상담 의도에 따라 하나의 상담에서도 조회·이체·연장·입금·상환·해지 등
여러 금융업무가 연결될 수 있으며, 업무영역 간 공통적인 처리행위도 광범위하게 존재한다.

따라서 FPG의 데이터 접근 구조 역시 단순한 부서 또는 업무 파트별 고정 할당만으로 설계하기보다,

**Role + 상담 Purpose + 실제 수행 Action**

을 함께 판단하여 상담 과정에서 필요한 데이터 접근범위를 동적으로 결정하는 구조가 적합하다는
분석적 근거를 확보하였다.
